Exercıcios de NumPy 19

```python
import numpy as np

array1 = np.random.randint(1, 101, 20)
array2 = np.random.randint(1, 101, 20)
produto = np.dot(array1, array2)
norma1 = np.linalg.norm(array1)
norma2 = np.linalg.norm(array2)
coseno = produto / (norma1 * norma2)
angulo = np.arccos(coseno)
angulo = np.degrees(angulo)

print(angulo)

```

Exercıcios de SciPy 3

```python
from scipy import optimize

def f(x):
    return x**4 - 3*x**3 + 2

for tentativa in [-5, -2, -1, 0, 1, 2, 3, 4, 5]:
    raiz = optimize.root(f, tentativa)
    print(tentativa, raiz.x)
```

Exercıcios de SciPy 4

```python
from scipy import optimize

def f(x):
    return -x**2 + 4*x + 1

def negativa(x):
    return -f(x)

resultado = optimize.minimize(negativa, 1)

print(resultado.x)
print(f(resultado.x))
```

Exercıcios de SciPy 9

```python
import numpy as np
from scipy.optimize import curve_fit

def curva(x, a, b, c):
    return a*x**2 + b*x + c

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 7, 8])

parametros, _ = curve_fit(curva, x, y)

print(parametros)
```

Exercıcios de SciPy 10

```python
import numpy as np
from scipy import integrate

def f(x, y):
    return x*y + x**2

resultado, erro = integrate.dblquad(
     f,
     0, 2,
     lambda x: 0,
     lambda x: 1
)

print(resultado)
print(erro)
```
