```python
def validar_cpf(cpf):
    cpf = cpf.replace(".", "")
    cpf = cpf.replace("-", "")

    if len(cpf) != 11:
        print("cpf invalido")
        return

    soma = 0

    for i in range(9):
        soma += int(cpf[i]) * (10 - i)

    resto = soma % 11

    if resto < 2:
        primeiro_digito = 0
    else:
        primeiro_digito = 11 - resto

    soma = 0

    for i in range(10):
        soma += int(cpf[i]) * (11 - i)

    resto = soma % 11

    if resto < 2:
        segundo_digito = 0
    else:
        segundo_digito = 11 - resto

    if int(cpf[9]) and segundo_digito == int(cpf[10]):
        print("cpf valido")
    else:
        print("cpf invalido")

cpf = input("Digite o cpf: ")

validar_cpf(cpf)

```

```python
def bubble_sort(arr):
    n = len(arr)

    for i in range(n):
        for j in range(n - 1):
            if arr[j] > arr[j + 1]:
                 arr[j], arr[j + 1] = arr[j + 1], arr[j]

    return arr

lista = [5, 2, 8, 1, 3]

print(bubble_sort(lista))
```

```python
jogo = [
    ["", "", ""],
    ["", "", ""],
    ["", "", ""]
]


def mostrar_jogo(jogo):
    for linha in jogo:
        print(" | ".join(linha))


def verificar_vitoria(jogo, jogador):
    for linha in jogo:
        if linha[0] == jogador and linha[1] == jogador and linha[2] == jogador:
            return True

    for coluna in range(3):
        if jogo[0][coluna] == jogador and jogo[1][coluna] == jogador and jogo[2][coluna] == jogador:
            return True

    if jogo[0][0] == jogador and jogo[1][1] == jogador and jogo[2][2] == jogador:
        return True

    if jogo[0][2] == jogador and jogo[1][1] == jogador and jogo[2][0] == jogador:
        return True

    return False


jogador = "X"

while True:
    mostrar_jogo(jogo)

    posicao = int(input("Jogador " + jogador + ", escolha uma posição de 1 a 9: "))

    linha = (posicao - 1) // 3
    coluna = (posicao - 1) % 3

    if jogo[linha][coluna] != "":
        print("Essa posição já está ocupada!")
        continue

    jogo[linha][coluna] = jogador

    if verificar_vitoria(jogo, jogador):
        mostrar_jogo(jogo)
        print("Jogador", jogador, "venceu!")
        break

    ocupado = True

    for linha in jogo:
        for posicao in linha:
            if posicao == "":
                ocupado = False

    if ocupado:
        mostrar_jogo(jogo)
        print("Empate!")
        break

    if jogador == "X":
        jogador = "O"
    else:
        jogador = "X"
```

```python
def insertion_sort(lista):
    for i in range(1, len(lista)):
        chave = lista[i]
        j = i -1

        while j >= 0 and lista[j] > chave:
            lista[j + 1] = lista[j]
            j -= 1

        lista[j + 1] = chave

    return lista

lista = [1, 8, 5, 7, 9]

print(insertion_sort(lista))
```
